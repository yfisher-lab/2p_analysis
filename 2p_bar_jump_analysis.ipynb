```python
def plot_bump_shape_before_after_jump(jump_data, cell_type='EPG'):
    """
    Plot bump shape (mean ± SEM) before and after bar jump for a given cell type ('EPG' or 'delta7').

    Parameters:
        jump_data (dict): Dictionary containing bump shape data.
        cell_type (str): 'EPG' or 'delta7'
    """
    assert cell_type in ['EPG', 'delta7'], "cell_type must be either 'EPG' or 'delta7'"

    # Select data based on cell type
    df_before = jump_data[f'{cell_type}_bump_shape_before_bar_jump']
    df_after = jump_data[f'{cell_type}_bump_shape_after_bar_jump']

    # Compute mean and SEM
    mean_before = np.nanmean(df_before, axis=1)
    sem_before = np.nanstd(df_before, axis=1) / np.sqrt(df_before.shape[1])

    mean_after = np.nanmean(df_after, axis=1)
    sem_after = np.nanstd(df_after, axis=1) / np.sqrt(df_after.shape[1])

    # Plotting
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, len(mean_before) + 1), mean_before, color='dodgerblue', linewidth=3, label='Before bar jump')
    plt.fill_between(range(1, len(mean_before) + 1), mean_before - sem_before, mean_before + sem_before, 
                     color='dodgerblue', alpha=0.3)

    plt.plot(range(1, len(mean_after) + 1), mean_after, color='darkorange', linewidth=3, label='After bar jump')
    plt.fill_between(range(1, len(mean_after) + 1), mean_after - sem_after, mean_after + sem_after, 
                     color='darkorange', alpha=0.3)

    plt.title(f'{cell_type} Bump Shape Before vs After Bar Jump', fontsize=18, fontweight='bold', color='black')
    plt.xlabel("Glomerulus Index", fontsize=14)
    plt.ylabel("Normalized dF/F", fontsize=14)
    plt.xticks(fontsize=12)
    plt.yticks(fontsize=12)
    plt.legend(fontsize=12)
    plt.tight_layout()
    plt.show()
```
