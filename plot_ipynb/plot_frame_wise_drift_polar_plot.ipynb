```python
def plot_angle_polar(
    data_all,
    column,
    *,
    flytrial_column="FlyTrial",
    color_by_fly=True,
    num_bins=30,
    rlim=15,
    title=None,
    save_path=None,

    # ===== NEW OPTIONAL COLOR ARGUMENTS =====
    bar_color="#ec1c24",       # default red
    edge_color="black",        # default black outline
    bar_alpha=0.90,            # for non-fly plotting

    fly_alpha=0.85             # for fly-stacked bars
):
    """
    Polar histogram for angle differences.
    Optional color control for bars and edges.
    """

    import numpy as np
    import matplotlib.pyplot as plt

    # ============================================================
    # Extract angles
    # ============================================================
    df = data_all['output_df_pooled']
    angles_deg = df[column].dropna()

    angles_rad = np.deg2rad(angles_deg)
    angles_rad = np.where(angles_rad > np.pi, angles_rad - 2*np.pi, angles_rad)

    # ============================================================
    # Fly extraction
    # ============================================================
    fly_ids_full = df[flytrial_column].dropna().loc[angles_deg.index]
    fly_ids = fly_ids_full.apply(lambda s: s.split('-')[0])
    unique_flies = fly_ids.unique()

    # Colormap
    fly_colors = plt.cm.tab20(np.linspace(0, 1, len(unique_flies)))
    fly_color_map = {fly: fly_colors[i] for i, fly in enumerate(unique_flies)}

    # ============================================================
    # Polar axis
    # ============================================================
    fig = plt.figure(figsize=(12, 8))
    ax = plt.subplot(111, projection='polar')

    edges = np.linspace(-np.pi, np.pi, num_bins + 1)
    centers = 0.5 * (edges[:-1] + edges[1:])
    widths = np.diff(edges)

    counts, _ = np.histogram(angles_rad, bins=edges)

    # ============================================================
    # (A) Single-color histogram
    # ============================================================
    if not color_by_fly:
        ax.bar(
            centers, counts,
            width=widths,
            bottom=0,
            align='center',
            color=bar_color,        # 👈 NEW OPTION
            edgecolor=edge_color,   # 👈 NEW OPTION
            linewidth=1.2,
            alpha=bar_alpha,        # 👈 NEW OPTION
            label="# of Events"
        )

    # ============================================================
    # (B) Fly-colored stacked histogram
    # ============================================================
    else:
        fly_bin_counts = {fly: np.zeros_like(counts) for fly in unique_flies}

        # Bin assignment
        bin_idx = np.digitize(angles_rad, bins=edges) - 1
        bin_idx = np.clip(bin_idx, 0, len(counts) - 1)

        for i, b in enumerate(bin_idx):
            fly = fly_ids.iloc[i]
            fly_bin_counts[fly][b] += 1

        bottom_vals = np.zeros_like(counts)

        for fly in unique_flies:
            ax.bar(
                centers,
                fly_bin_counts[fly],
                width=widths,
                bottom=bottom_vals,
                align='center',
                color=fly_color_map[fly],
                edgecolor=edge_color,   # 👈 still consistent
                linewidth=0.8,
                alpha=fly_alpha,        # 👈 NEW OPTION
                label=fly
            )
            bottom_vals += fly_bin_counts[fly]

    # ============================================================
    # Polar axis formatting
    # ============================================================
    ax.set_thetalim(-np.pi, np.pi)
    ax.set_xticks([-np.pi, -3*np.pi/4, -np.pi/2, -np.pi/4,
                   0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi])
    ax.set_xticklabels([r'', r'-135°', r'-90°', r'-45°', r'0°',
                        r'45°', r'90°', r'135°', r'±180°'])

    ax.set_rticks([0,5,10,15,20,25])
    ax.set_rlim(0, 15)

    if title:
        ax.set_title(title, fontsize=16)

    ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1.15))
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        plt.close(fig)
    else:
        plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.stats import circmean, circstd
from scipy.spatial.distance import euclidean


# ============================================================
# Helper: convert degrees → radians in [-pi, pi]
# ============================================================
def wrap_deg_to_rad(x):
    r = np.deg2rad(x)
    return np.where(r > np.pi, r - 2*np.pi, r)


# ============================================================
# Helper: Watson U² two-sample test (nonparametric)
# ============================================================
def watson_U2_test(sample1, sample2):
    """
    Nonparametric circular test comparing two distributions.
    Returns the U² statistic; significance threshold depends on critical values.
    """

    # Combine and sort all samples
    data = np.concatenate([sample1, sample2])
    n1, n2 = len(sample1), len(sample2)
    N = n1 + n2

    # sort all angles
    order = np.argsort(data)
    data_sorted = data[order]

    # indicator arrays for group membership
    ind1 = np.isin(order, np.argsort(sample1))
    
    # ECDFs
    F1 = np.cumsum(ind1) / n1
    F2 = np.cumsum(~ind1) / n2

    # Watson U2 formula
    U2 = np.sum((F1 - F2)**2) - (n1 - n2)**2 / (12 * N**2)
    return U2


# ============================================================
# Helper: permutation test comparing mean resultant vectors
# ============================================================
def circular_permutation_test(rad1, rad2, n_perm=5000):
    """
    Tests whether two circular samples differ by comparing 
    the length of the resultant vector difference under permutation.
    """
    rad1 = np.asarray(rad1)
    rad2 = np.asarray(rad2)
    n1, n2 = len(rad1), len(rad2)

    # observed statistic: difference between mean vectors
    v1 = np.array([np.mean(np.cos(rad1)), np.mean(np.sin(rad1))])
    v2 = np.array([np.mean(np.cos(rad2)), np.mean(np.sin(rad2))])
    obs = euclidean(v1, v2)

    # pooled data
    pooled = np.concatenate([rad1, rad2])

    # permutation procedure
    count = 0
    for _ in range(n_perm):
        perm = np.random.permutation(pooled)
        p1 = perm[:n1]
        p2 = perm[n1:]

        v1p = np.array([np.mean(np.cos(p1)), np.mean(np.sin(p1))])
        v2p = np.array([np.mean(np.cos(p2)), np.mean(np.sin(p2))])
        stat = euclidean(v1p, v2p)

        if stat >= obs:
            count += 1

    p_value = (count + 1) / (n_perm + 1)
    return obs, p_value


# ============================================================
# Main function: compare two datasets
# ============================================================
def compare_angle_distributions(
    data_all_1,
    data_all_2,
    *,
    angle_column="angle difference start/5s",
    duration_column="Period_duration",
    min_duration=5.0,
    subsample=True,
    duration_weighting=True,
    n_perm=5000
):
    """
    Compares circular angle distributions between dataset 1 and dataset 2.
    Controls for:
      - different sample sizes (via subsampling)
      - different stop durations (via weighting)

    Returns:
        dict of results including:
            - descriptive stats
            - Watson U² statistic
            - permutation test p-value
    """

    # ===========================
    # Extract valid angles
    # ===========================
    df1 = data_all_1['output_df_pooled']
    df2 = data_all_2['output_df_pooled']

    # filter by minimum stop duration
    df1 = df1[df1[duration_column] >= min_duration]
    df2 = df2[df2[duration_column] >= min_duration]

    ang1 = wrap_deg_to_rad(df1[angle_column].dropna().values)
    ang2 = wrap_deg_to_rad(df2[angle_column].dropna().values)

    dur1 = df1[duration_column].values
    dur2 = df2[duration_column].values

    # ===========================
    # Duration weighting
    # ===========================
    # Longer stops → more repetitions of the angle measurement
    if duration_weighting:
        w_ang1 = np.repeat(ang1, np.maximum(1, (dur1 / min_duration)).astype(int))
        w_ang2 = np.repeat(ang2, np.maximum(1, (dur2 / min_duration)).astype(int))
    else:
        w_ang1 = ang1
        w_ang2 = ang2

    # ===========================
    # Equalize sample sizes
    # ===========================
    if subsample:
        n = min(len(w_ang1), len(w_ang2))
        idx1 = np.random.choice(len(w_ang1), n, replace=False)
        idx2 = np.random.choice(len(w_ang2), n, replace=False)
        w_ang1 = w_ang1[idx1]
        w_ang2 = w_ang2[idx2]

    # ===========================
    # Statistics
    # ===========================
    mean1 = circmean(w_ang1, high=np.pi, low=-np.pi)
    mean2 = circmean(w_ang2, high=np.pi, low=-np.pi)
    r1 = 1 - circstd(w_ang1, high=np.pi, low=-np.pi)**2
    r2 = 1 - circstd(w_ang2, high=np.pi, low=-np.pi)**2

    # nonparametric Watson U² test
    U2 = watson_U2_test(w_ang1, w_ang2)

    # permutation test on mean vectors
    obs, p_perm = circular_permutation_test(w_ang1, w_ang2, n_perm=n_perm)

    return {
        "n1_original": len(ang1),
        "n2_original": len(ang2),
        "n1_weighted": len(w_ang1),
        "n2_weighted": len(w_ang2),
        "circular_mean_1": mean1,
        "circular_mean_2": mean2,
        "resultant_length_1": r1,
        "resultant_length_2": r2,
        "Watson_U2": U2,
        "permutation_statistic": obs,
        "permutation_p_value": p_perm
    }
```

```python
results = compare_angle_distributions(
    data_all_1,
    data_all_2,
    angle_column="angle difference start/5s",
    duration_column="Period_duration",
    min_duration=5.0,
    subsample=True,
    duration_weighting=True,
    n_perm=5000
)

print(results)
```
